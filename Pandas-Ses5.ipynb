```python
movies_data = pd.merge(movies_financials, languages, on = "language_id", how = "left")
```

```python
movies_data.sample(4)
```

```python
movies_data.columns = movies_data.columns.str.replace("name" , "language")
```

```python
movies_data.sample(3)
```

```python
movies_data["revenue_millions"] = movies_data.apply(
    lambda row :
    row["revenue"] *1000
    if row["unit"]=="Billions"
    else row["revenue"],
    axis = 1
)
```

```python
movies_data["budget_millions"] = movies_data.apply(
    lambda row :
    row["budget"] *1000
    if row["unit"]=="Billions"
    else row["budget"],
    axis = 1
)
```

```python
movies_data["unit"].value_counts()
```

```python
movies_data["profit_millions"] = movies_data["revenue_millions"] - movies_data["budget_millions"]
```

```python
movies_data.head(3)
```

```python
industry_map = {                         #"map" is used to change the name of the column
    "Hollywood" : "International",
    "Bollywood" : "Indian"
}
```

```python
movies_data["industry_type"] = movies_data["industry"].map(industry_map)
```
